```python
import re
import numpy as np
import matplotlib.pyplot as plt

import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow import keras
from tensorflow.keras import layers

```

```python
text = "deep learning is a part of machine learning. machine learning uses data to train models.neural networks are models used in deep learning. a neural network has layers of neurons.the model learns patterns from data. natural language processing uses deep learning models to understand text.word embeddings represent words as vectors. the cbow model predicts a target word from its context words.skip gram predicts context words from a target word. embeddings capture the meaning of words in vectors."

tokenizer=Tokenizer()
tokenizer.fit_on_texts([text])

words=tokenizer.word_index
vocab_size=len(words)+1
```

```python
seq=tokenizer.texts_to_sequences([text])[0]

x=[]
y=[]

for i in range(1,len(seq)-1):
    x.append([seq[i-1],seq[i+1]])
    y.append(seq[i])
    
x=np.array(x)
y=np.array(y)

print(x.shape)
print(len(seq))
print(y.shape)
```

```python
model=keras.Sequential([
    layers.Embedding(vocab_size,10),
    layers.Lambda(lambda x: tf.reduce_mean(x,axis=1)),
    layers.Dense(vocab_size, activation="softmax")
])
model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history=model.fit(   
    x,     
    y,     
    epochs=100,     
    verbose=1 
)
```

```python
test = np.array([[words[ 'a' ], words[ 'learning' ]]])
prediction = model.predict(test, verbose=0)
index = np.argmax(prediction) 
predicted_word = tokenizer.index_word[index] 
print( "Predicted word:" , predicted_word)
```
